```python
# import the data
import json
```

```python
# Load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("sample/data.json");
print(data)
print(type(data))
```

# Cleaning and Structuring Data
1. Data is inconsistent => mixed type
2. Missing Data
3. Duplicate Data

```python
# Cleaning and Structure the Data
def clean_data(data):
    text_to_number = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()
    for user in data:
        # Clean rating
        raw_rating = str(user["rating"]).strip().lower()
        if (raw_rating in text_to_number):
            raw_rating = text_to_number[raw_rating]
        user["rating"] = raw_rating

        # clean name
        user["name"] = user["name"].strip()

        # Clean feedback
        user["feedback"] = user["feedback"].strip()

        # handle Missing values
        raw_age = user.get("age")
        if (raw_age == None):
            user["age"] = None

        # Deduplication
        if (user["name"] in unique_users):
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
```

```python
data = clean_data(data);
for user in data:
    print(user)
```

# Assignment 2
Getting meaningful insights from data

1. What's avg rating?
2. What % of users gave poor rating (<3)?

```python
# Get meaningful insights from data
def get_insights(data):
    # get the avg rating
    total_rating = 0
    for user in data:
        total_rating += float(user["rating"]);

    print(f"avg rating = {total_rating / len(data)}")

    # Percentage of users who gave poor rating
    poor_rating_count = 0
    for user in data:
        if (float(user["rating"]) < 3):
            poor_rating_count += 1

    print(f"Percentage of users who gave poor rating = {(poor_rating_count / len(data)) * 100}%")
```

```python
get_insights(data)
```

# Assignment 3
Build a product **recomendation feature** (Rule based, not ML)

1. If user rating >= 4, Recommend same brand products.
2. If user rating < 4, Recommend different brand products.

```python
# Build recommendation Feature

def get_recommendations(data):
    recommendations = []

    for user in data:
        current_recommendation = {}
        current_recommendation["name"] = user["name"]
        if (float(user["rating"]) >= 4):
            current_recommendation["brand"] = "Apple"
        else:
            current_recommendation["brand"] = "Samsung"
        recommendations.append(current_recommendation)

    return recommendations
```

```python
get_recommendations(data)
```
